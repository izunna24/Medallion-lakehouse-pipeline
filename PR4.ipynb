```python
%%writefile /content/drive/MyDrive/my_project_4

# I introduced Executor Sizing Formula for the pipeline
# With respect to my laptop as follows;;
# Processor = 6 Core(s), 8 Logical Processor(s)
# RAM = 8GB, 7.6GB usable memory
# Node = 1 Node
# Pipeline File Size = aprox. 30MB
# My laptop is basically for Data Engineering traing, not for production work

# {Import core Spark components,
# Delta Lake integration,
# and standard utility modules.}
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from delta.tables import DeltaTable
from datetime import datetime
from pyspark.sql.types import *
from delta import configure_spark_with_delta_pip
import getpass, time, os, shutil
from pyngrok import ngrok
from pyspark.sql.window import Window

# Clean up any stale directory paths on the local file system.
# This prevents schema drift/conflicts when testing changes to the pipeline.
paths_to_clean = [
    "/path/to/lakehouse/bronze/1/orders",
    "/path/to/lakehouse/bronze/2/orders",
    "/path/to/lakehouse/silver/1/orders_quarantine",
    "/path/to/lakehouse/silver/2/orders_quarantine",
    "/path/to/lakehouse/MLops/orders_quarantine",
    "/path/to/lakehouse/silver/orders",
    "/path/to/lakehouse/gold/agg",
    "/path/to/lakehouse/MLops/orders",
    "/path/to/lakehouse/good_rst/orders",
    "/path/to/lakehouse/clean_rst/orders",
    "/path/to/lakehouse/ML/lr_pipeline",
    "/path/to/lakehouse/predictions/delta_ml"
]

for p in paths_to_clean:
    shutil.rmtree(p, ignore_errors=True)
print("Local stale data cleared!")

from pyspark.sql import SparkSession

builder = (
    SparkSession.builder
    .appName("MedallionLakehousePipeline")
    # Preferred for 30 MB data on an 8 GB RAM laptop
    .master("local[6]")
    .config("spark.driver.memory", "3g")
    .config("spark.executor.instances", "1") # ignore or set to 1
    .config("spark.executor.memoryOverhead", "1")
    .config("spark.sql.shuffle.partitions", "12")
    # .config("spark.executor.memory", "3g")   # ignored in local mode
    # .config("spark.executor.cores", "2")     # ignored in local mode
    # ------Delta-Spark comfigurations ------
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
    # ----- Spark UI comfiguration -
    .config("spark.ui.port", "4050")
    # ---- extra useful configs ----
    .config("spark.sql.adaptive.enabled", "true")
    .config("spark.sql.adaptive.skewJoin.enabled", "true")
    .config("spark.sql.adaptive.coalescePartitions.enabled", "true")
    .config("spark.serializer", "org.apache.spark.serializer.KryoSerializer")
    .config("spark.sql.files.maxPartitionBytes", "16m")
    # .config("spark.databricks.delta.optimizeWrite.enabled", "true")
    # .config("spark.databricks.delta.autoCompact.enabled", "true")
)
spark = configure_spark_with_delta_pip(
    builder, extra_packages=["io.delta:delta-spark_2.13:4.0.0"]
).getOrCreate()

spark.sparkContext.setLogLevel("WARN")

# Check if SparkSession exists and what UI port it is using
try:
    print("Spark UI port:", spark.sparkContext.getConf().get("spark.ui.port"))
    print("Spark UI enabled:", spark.sparkContext.getConf().get("spark.ui.enabled", "true"))
    print("SparkSession is active")
except Exception as e:
    print("SparkSession is NOT active:", e)

# 1. Gracefully disconnect every active tunnel
ngrok.kill()
time.sleep(1)
print("ngrok initial tunnels killed.")


authtoken = getpass.getpass()
ngrok.set_auth_token(authtoken)

public_url = ngrok.connect(4050).public_url

# Display connection details and external access URL for the Spark Web UI
print(f"\nSpark UI: http://localhost:4050")
print("-" * 10)
print(f"👉 Click this link to open your Spark UI: \n{public_url}")
print("-" * 10)
print("Keep this cell running while you use the Spark UI.")

print(f"Spark UI: http://localhost:4050")

# Lakehouse Medallion Architecture Paths
bronze_path1 =  "/path/to/lakehouse/bronze/1/orders"
bronze_path2 =  "/path/to/lakehouse/bronze/2/orders"
quarantine_path1 = "/path/to/lakehouse/silver/1/orders_quarantine"
quarantine_path2 = "/path/to/lakehouse/silver/2/orders_quarantine"
quarantine_ml_path = "/path/to/lakehouse/MLops/orders_quarantine"
silver_path = "/path/to/lakehouse/silver/orders"
gold_path = "/path/to/lakehouse/gold/agg"
ml_path = "/path/to/lakehouse/MLops/orders"
good_rst = "/path/to/lakehouse/good_rst/orders"
clean_rst = "/path/to/lakehouse/clean_rst/orders"
model_path = "/path/to/lakehouse/ML/lr_pipeline"
delta_ml = "/path/to/lakehouse/predictions/delta_ml"

# File names for raw input data batches and inference inputs
raw_day1 = "orders_day1_skewed.csv"
raw_day2 = "orders_day2_5.csv"
raw_prediction = "orders_prediction.csv"
# Relative directory path where raw CSV files are located
head_data = "data/raw/"


# 1. Clear old checkpoints first
for tier in ["bronze/st_1", "bronze/st_2", "ml_st"]:
    path = f"stream/checkpoint/for/{tier}"
    if os.path.exists(path):
        shutil.rmtree(path, ignore_errors=True)

# 2. Re-create clean checkpoint directories
for tier in ["bronze/st_1", "bronze/st_2", "ml_st"]:
    os.makedirs(f"stream/checkpoint/for/{tier}", exist_ok=True)

bronze_st1 = "stream/checkpoint/for/bronze/st_1"
bronze_st2 = "stream/checkpoint/for/bronze/st_2"
ml_st = "stream/checkpoint/for/ml_st"

def explain_and_log(df, name: str):
    print(f"\n===== EXPLAIN: {name} =====")
    df.explain("formatted")
    print(f"→ Check Spark UI (SQL tab) for '{name}'\n")

def dataframe_size(df):
    """Estimate DataFrame size in bytes using Spark's plan statistics."""
    # Check if the DataFrame is already cached
    was_cached = df.is_cached

    if not was_cached:
        df.cache()
    df.count()  # force materialization

    # Access the underlying JVM SparkSession
    # Read `sizeInBytes` from `optimizedPlan`.
    try:
        # Works on most Spark 3.x versions
        size_bytes = (spark._jsparkSession.sessionState()
            .executePlan(df._jdf.queryExecution().logical())
            .optimizedPlan().stats().sizeInBytes()
        )
    # where `executePlan` requires an explicit execution mode argument.
    except Exception:
        # Spark ≥ 3.2 often requires the mode argument
        size_bytes = (spark._jsparkSession.sessionState()
            .executePlan(df._jdf.queryExecution().logical(),
                df._jdf.queryExecution().mode())
            .optimizedPlan().stats().sizeInBytes()
        )

    # Only unpersist if the function itself initiated the cache
    if not was_cached:
      df.unpersist()

    return size_bytes


# Bronze Layer
def bronze_stream(raw_path: str, ingest_path: str, checkpoint_path: str, parent_dir):
  full_path = os.path.join(parent_dir, raw_path)
  raw_df = \
        spark.read.option("header", True) \
        .csv(full_path)
  print(f"data row count: {raw_df.count()}")
  raw_df.printSchema()
  raw_df.show(5, truncate=False)

  def process_batch(bronze_batch_df, batch_id: int):
    if bronze_batch_df.isEmpty():
        return

    if not DeltaTable.isDeltaTable(spark, ingest_path):
        # First batch: create the table
        (bronze_batch_df
            .write
            .format("delta")
            .mode("append")
            .option("mergeSchema", "true")
            .partitionBy("order_date")
            .save(ingest_path)
        )
        print(f"Batch {batch_id}: Created new Delta table")
    else:
        # Subsequent batches: MERGE
        bronze_tbl = DeltaTable.forPath(spark, ingest_path)

        (bronze_tbl.alias("target")
            .merge(
                bronze_batch_df.alias("source"),
                "target.order_id = source.order_id AND target.order_date = source.order_date"
            )
            .whenMatchedUpdateAll()
            .whenNotMatchedInsertAll()
            .execute()
        )
        print(f"Batch {batch_id}: MERGE completed")

  # --- Schema: schema-on-read, all strings (Bronze never coerces) ---
  raw_schema = """
    order_id STRING,
    line_item_seq STRING,
    customer_id STRING,
    product_id STRING,
    region STRING,
    order_date STRING,
    status STRING,
    quantity STRING,
    unit_price STRING
   """
  # --- Streaming read ---
  bronze_stream_df = (
    spark.readStream
    .format("csv")
    .option("header", "true")
    .schema(raw_schema)
    .option("maxFilesPerTrigger", 1)   # one file per micro-batch
    .option("pathGlobFilter", raw_path)
    .load(parent_dir)
    .withColumn("_source_file", lit(os.path.basename(raw_path)))
  )

  query = (
    bronze_stream_df.writeStream
    .foreachBatch(process_batch)
    .option("checkpointLocation", checkpoint_path)
    .trigger(availableNow=True)
    .start()
  )

  query.awaitTermination()

  # Optional Z-ORDER (helps later reads)
  DeltaTable.forPath(spark, ingest_path).optimize().executeZOrderBy("order_id")

t0 = time.time()
bronze_stream(raw_day1, bronze_path1, bronze_st1, head_data)
bronze_stream(raw_day2, bronze_path2, bronze_st2, head_data)
bronze_stream(raw_prediction, ml_path, ml_st, head_data)
t1 = time.time()
print(f"Bronze layer Run_time: {t1 - t0:.2f}s")
print("Bronze Stream is successful")

bronze_df1 = spark.read.format("delta").load(bronze_path1)
bronze_df2 = spark.read.format("delta").load(bronze_path2)
ml_df = spark.read.format("delta").load(ml_path)
print(f"Row Count: B1: {bronze_df1.count()}, B2: {bronze_df2.count()}, Ml: {ml_df.count()}")

size0 = dataframe_size(bronze_df1)
size1 = dataframe_size(bronze_df2)
size2 = dataframe_size(ml_df)
print(f"bronze_df1 size: {size0 / (1024):.2f} KiB")
print(f"bronze_df2 size: {size1 / (1024):.2f} KiB")
print(f"ml_df size: {size2 / (1024):.2f} KiB")

# Silver Layer
def schema_audit(ingest_df, ingest_path):
    # Audit for not trimed values
    string_cols = [
    f.name for f in ingest_df.schema.fields
    if isinstance(f.dataType, StringType)
  ]

    whitespace_counts_df = ingest_df.select([
    sum(
        when(length(col(c)) != length(trim(col(c))), 1).otherwise(0)
    ).alias(c) for c in string_cols
    ])

    print("whitespace content")
    whitespace_counts_df.show()

    schema_df = ingest_df.select(
        col("order_id").cast(StringType()),
        col("customer_id").cast(StringType()),
        col("product_id").cast(StringType()),
        col("quantity").cast(IntegerType()),
        col("unit_price").cast(DoubleType()),
        col("region").cast(StringType()),
        col("status").cast(StringType()),
        col("order_date").try_cast(DateType()),   # malformed dates → null
        col("line_item_seq").cast(IntegerType()),
    )

    schema_df = schema_df \
       .withColumn("_source_file",
        lit(os.path.basename(ingest_path)).cast(StringType())) \
        .withColumn("ingest_date",
                lit(current_date()))

    schema_df.printSchema()

    # Audit for null values
    schema_df.select([
        sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
        for c in schema_df.columns
    ]).show()

    # Audit for numerics less than and/or equal zero
    numeric_types = (IntegerType, DoubleType)

    # 1. Collect numeric column names (small metadata list – unavoidable)
    numeric_cols = [
    field.name
    for field in schema_df.schema.fields
    if isinstance(field.dataType, numeric_types)
    ]

    print("Numeric columns:", numeric_cols)

    # One-row DataFrame with the count of ≤ 0 values per numeric column
    counts_df = schema_df.select([
    sum(when(col(c) <= 0, 1).otherwise(0)).alias(c)
    for c in numeric_cols
    ])

    counts_df.show()

    # Explain
    explain_and_log(schema_df, "schema_audit result")

    return schema_df

bronze_df1.cache()
bronze_df2.cache()
ml_df.cache()
schema_audit(bronze_df1, bronze_path1)
schema_audit(bronze_df2, bronze_path2)
schema_audit(ml_df, ml_path)
bronze_df1.unpersist()
bronze_df2.unpersist()
ml_df.unpersist()

print("Bronze_df1 Delta history:")
DeltaTable.forPath(spark, bronze_path1).history().show(truncate=False)
print("Bronze_df2 Delta history:")
DeltaTable.forPath(spark, bronze_path2).history().show(truncate=False)
print("ml_df Delta history:")
DeltaTable.forPath(spark, ml_path).history().show(truncate=False)


def quality_checks(df):
    # Define primary business key columns and unique transactional event
    business_cols = ["order_id", "customer_id", "product_id", "quantity",
                      "unit_price", "region", "status", "order_date",]

    window_spec = Window.partitionBy(*business_cols)

    sort_df = df.withColumn("row_signature_count",
    count("*").over(window_spec))

    # Each rule is: (rule_name, condition_column_expr, description)
    q_rules= [
    (
        "order_id_not_null",
        col("order_id").isNotNull(),
        "order_id must not be null"
    ),
    (
         "customer_id_not_null",
        col("customer_id").isNotNull(),
        "customer_id must not be null"
    ),
    (
         "product_id_not_null",
        col("product_id").isNotNull(),
        "product_id must not be null"
    ),
    (
        "quantity_greater_than_zero_&_not_null",
        col("quantity").isNotNull() & (col("quantity") > 0),
        "quantity must be greatter than zero and not null"
    ),
    (
        "unit_price_greater_than_zero_&_not_null",
        col("unit_price").isNotNull() & (col("unit_price") > 0),
        "unit_price must be greatter than zero and not null"
    ),
    (
         "region_not_null",
        col("region").isNotNull(),
        "region must not be null"
    ),
    (
        "status_valid",
        col("status").isin("PLACED", "PROCESSING", "SHIPPED", "DELIVERED", "CANCELLED"),
        "Status must be PLACED, PROCESSING, SHIPPED, DELIVERED or CANCELLED"
    ),
    (
        "order_date_not_null",
        col("order_date").isNotNull(),
        "order_date must not be null"
    ),
    (
        "row_signature_count_equals_1",
        (col("row_signature_count") == 1),
        "row_signature_count must be 1"
    ),
    ]

    # Build a column for each rule — True if passed, False if failed
    df_checked = sort_df
    for rule_name, condition, _ in q_rules:
        df_checked = df_checked.withColumn(
            f"_check_{rule_name}",
            when(condition, lit(True)).otherwise(lit(False))
        )

    # Collect names of failed checks into an array column
    failed_checks_col = array(*[
        when(col(f"_check_{rule_name}") == False, lit(rule_name))
        .otherwise(lit(None))
        for rule_name, _, _ in q_rules
    ])

    df_checked = df_checked.withColumn(
        "_failed_rules_raw", failed_checks_col
    )

    # Filter nulls out of the array to get only failed rule names
    from pyspark.sql.functions import array_compact
    df_checked = df_checked.withColumn(
        "_failed_rules",
        array_compact(col("_failed_rules_raw"))
    )

    # Drop intermediate columns
    check_cols = [f"_check_{r}" for r, _, _ in q_rules] + ["_failed_rules_raw"]
    df_checked = df_checked.drop(*check_cols)

    # Split into good and bad
    from pyspark.sql.functions import size
    good_df = df_checked \
        .filter(size(col("_failed_rules")) == 0) \
        .drop("_failed_rules", "row_signature_count")

    good_df.show(5, truncate=False)

    bad_df = df_checked \
        .filter(size(col("_failed_rules")) > 0) \
        .withColumn("_quarantined_at", current_timestamp()) \
        .drop("row_signature_count")

    bad_df.show(5)

    return good_df, bad_df

t2 = time.time()
check_day1 = schema_audit(bronze_df1, bronze_path1).cache()
check_day2 = schema_audit(bronze_df2, bronze_path2).cache()
check_ml = schema_audit(ml_df, ml_path).cache()

check_day1.unpersist()
check_day2.unpersist()
check_ml.unpersist()

good_day1, bad_day1 = quality_checks(check_day1)
good_day2, bad_day2 = quality_checks(check_day2)
good_ml, bad_ml = quality_checks(check_ml)

size4 = dataframe_size(good_day1)
size5 = dataframe_size(bad_day1)
size6 = dataframe_size(good_day2)
size7 = dataframe_size(bad_day2)
size8 = dataframe_size(good_ml)
size9 = dataframe_size(bad_ml)
print(f"good_day1 size: {size4 / (1024):.2f} KiB")
print(f"bad_day1 size: {size5 / (1024):.2f} KiB")
print(f"good_day2 size: {size6 / (1024):.2f} KiB")
print(f"bad_day2 size: {size7 / (1024):.2f} KiB")
print(f"good_ml size: {size8 / (1024):.2f} KiB")
print(f"bad_ml size: {size9 / (1024):.2f} KiB")

print(f"count rows: G1: {good_day1.count()}, G2: {good_day2.count()}, G_ml: {good_ml.count()}")
print(f"count rows: B1: {bad_day1.count()}, B2: {bad_day2.count()}, B_ml: {bad_ml.count()}")


def quarantine(qua_df, qua_path):
  if qua_df.limit(1).count() > 0:
    qua_df.write.format("delta") \
    .mode("append").option("mergeSchema", "true") \
    .partitionBy("order_date") \
    .save(qua_path)
  print(f"quarantine writen to {qua_path}")

quarantine(bad_day1, quarantine_path1)
quarantine(bad_day2, quarantine_path2)
quarantine(bad_ml, quarantine_ml_path)
t3 = time.time()
print(f"Silver DQ + quarantine Run_time: {t3 - t2:.2f}s")

q1_df = spark.read.format("delta").load(quarantine_path1)
q2_df = spark.read.format("delta").load(quarantine_path2)
q_ml = spark.read.format("delta").load(quarantine_ml_path)
print(f"quarantine count: q1={q1_df.count()}, q2={q2_df.count()}, q_ml={q_ml.count()}")


# loops with same base path "stream/checkpoint/for/"
# Consolidated into one clean block.
for tier in ["silver/st", "gold/st"]:
    path = f"stream/checkpoint/for/{tier}"
    if os.path.exists(path):
        shutil.rmtree(path, ignore_errors=True)
    os.makedirs(f"stream/checkpoint/for/{tier}", exist_ok=True)

silver_st = "stream/checkpoint/for/silver/st"
gold_st = "stream/checkpoint/for/gold/st"

# Write good data first before streaming it.
print("Writing good data to staging path for silver to consume...")

# Good data A
distinct_dates_d1 = [r.order_date for r in good_day1.select("order_date")
.distinct().orderBy("order_date").collect()]
for d in distinct_dates_d1:
    good_day1.filter(col("order_date") == d) \
        .coalesce(1).write.format("delta") \
        .mode("append").option("mergeSchema", "true") \
        .option("delta.autoOptimize.optimizeWrite", "false") \
        .partitionBy("order_date").save(good_rst)

for d in distinct_dates_d1:
    active_files = spark.read.format("delta").load(good_rst) \
        .filter(col("order_date") == d).inputFiles()
    print(d, "→", len(active_files), "active file(s)")

# Good data B
distinct_dates_d2 = [r.order_date for r in good_day2.select("order_date")
.distinct().orderBy("order_date").collect()]
for d in distinct_dates_d2:
    good_day2.filter(col("order_date") == d) \
        .coalesce(1).write.format("delta") \
        .mode("append").option("mergeSchema", "true") \
        .option("delta.autoOptimize.optimizeWrite", "false") \
        .partitionBy("order_date").save(good_rst)

for d in distinct_dates_d2:
    active_files = spark.read.format("delta").load(good_rst) \
        .filter(col("order_date") == d).inputFiles()
    print(d, "→", len(active_files), "active file(s)")

for root, dirs, files in os.walk(good_rst):
    parquet_files = [f for f in files if f.endswith(".parquet")]
    if parquet_files:
        print(root, "→", len(parquet_files), "file(s)")

# New signature is simple and correct: source -> target -> checkpoint
def silver_stream(clean_source_path, silver_target_path, checkpoint_path):
    """
    clean_source_path=good_rst:     the 2 file sources for the mearge
    silver_target_path=silver_path: the final path after merge
    checkpoint_path=silver_st:      the streaming passage point
    """
    # FIXED: Single stream: read good_rst (delta) -> SCD2 MERGE into silver_path
    clean_stream_df = (
        spark.readStream
        .format("delta")
        .option("maxFilesPerTrigger", 1)
        .load(clean_source_path)
    )

    def merge_batch(source_df, batch_id: int):
        if source_df.isEmpty():
            return

        # Add SCD2 columns inside batch, not outside (outside is streaming DF)
        incoming_rows = source_df \
            .withColumn("is_current", lit(True)) \
            .withColumn("valid_from", col("ingest_date").cast(DateType())) \
            .withColumn("valid_to", lit(None).cast(TimestampType()))

        # second merge used same target variable after explain_and_log.
        if not DeltaTable.isDeltaTable(spark, silver_target_path):
            incoming_rows.write.format("delta").mode("append") \
                .option("mergeSchema", "true") \
                .partitionBy("order_date").save(silver_target_path)
            print(f"Batch {batch_id}: Created new silver table")
        else:
            target = DeltaTable.forPath(spark, silver_target_path)

            # Step 1: Expire (soft-close) old current record if status changed
            target.alias("tgt").merge(
                incoming_rows.alias("src"),
                condition="""
                    tgt.order_id = src.order_id
                    AND tgt.line_item_seq = src.line_item_seq
                    AND tgt.is_current = true
                    AND tgt.status != src.status
                """
            ).whenMatchedUpdate(set={
                "is_current": lit(False),
                "valid_to": current_timestamp()
            }).execute()

            # Step 2: Insert new version if it doesn't exist as current
            target.alias("tgt").merge(
                incoming_rows.alias("src"),
                condition="""
                    tgt.order_id = src.order_id
                    AND tgt.line_item_seq = src.line_item_seq
                    AND tgt.is_current = true
                """
            ).whenNotMatchedInsertAll().execute()

        print(f"Batch {batch_id}: SCD2 MERGE completed")

    query = (
        clean_stream_df.writeStream
        .foreachBatch(merge_batch)
        .option("checkpointLocation", checkpoint_path)
        .trigger(availableNow=True)
        .start()
    )
    query.awaitTermination()
    print("Silver Stream is successful")


good_day1 = good_day1.cache()
good_day2 = good_day2.cache()

t4 = time.time()
# FIXED CALL - correct arg order
silver_stream(
    clean_source_path=good_rst,
    silver_target_path=silver_path,
    checkpoint_path=silver_st
)

# Z-ORDER after all MERGEs
DeltaTable.forPath(spark, silver_path).optimize().executeZOrderBy(
             "order_id", "region")

print("silver layer Delta history:")
DeltaTable.forPath(spark, silver_path).history().show(truncate=False)

t5 = time.time()
print(f"Silver MERGE (SCD2) Run_time: {t5 - t4:.2f}s")

good_day1.unpersist()
good_day2.unpersist()

silver_df = spark.read.format("delta").load(silver_path)
print(f"Silver Count: {silver_df.count()}")
silver_df.show(5, truncate=False)

# Explain merge in silver layer
explain_and_log(silver_df, "Silver merge rows")

size10 = dataframe_size(silver_df)
print(f"silver_df size: {size10 / (1024):.2f} KiB")

silver_df.write.format("delta"). \
                mode("append").option("mergeSchema", "true") \
                .partitionBy("order_date").save(clean_rst)

# Get history using the DeltaTable API
delta_table = DeltaTable.forPath(spark, silver_path)
history_df = delta_table.history() \
.select("version", "timestamp", "operation").show(truncate=False)

def indicate_skew(skew_df, columns=None, top_n=10):
  if columns is None:
    columns = skew_df.columns

  for c in columns:
    print(f"\n===== Column: {c} =====")
    counts_df = skew_df.groupBy(c) \
          .count()

    # Get max, avg, min for skew detection
    stats = counts_df.agg(
            max("count").alias("max_count"),
            avg("count").alias("avg_count"),
            min("count").alias("min_count"),
            count("*").alias("distinct_keys")
        ).collect()[0]

    max_c = stats["max_count"]
    avg_c = stats["avg_count"]
    min_c = stats["min_count"]

    skew_ratio = max_c / avg_c if avg_c else 0

    print(f"max(count) = {max_c:,} | avg(count) = {avg_c:,.2f} | min(count) = {min_c:,} | distinct = {stats['distinct_keys']:,}")
    print(f"Skew Ratio (max/avg) = {skew_ratio:,.1f}x -> {'HIGH SKEW - SALT THIS' if skew_ratio > 10 else 'OK' if skew_ratio < 3 else 'MODERATE'}")

    # Top N heaviest keys
    counts_df.orderBy(desc("count")).show(top_n, truncate=False)


silver_df.cache()
good_ml.cache()

t6 = time.time()
# execute to indicate columns that are skewed
indicate_skew(silver_df)
indicate_skew(good_ml)
good_ml.unpersist()

# Gold layer
def gold_layer(salt_df, agg_path):
    n_salt = 10
    stragglers = ["9003", "9005", "9004", "9001", "9002"]  # 5 stragglers

    # 1. Salt ONLY the 5 straggler keys
    strag_salted = salt_df.withColumn(
        "salted_customer_id",
        when(
            col("customer_id").isin(stragglers),
            concat(
                col("customer_id").cast("string"),
                lit("_"),
                floor(rand() * n_salt).cast(IntegerType()).cast("string")
            )
        ).otherwise(col("customer_id").cast("string"))
    )

    strag_salted.show()

    print("=== Salted Key Distribution ===")
    # Fix: Filter before groupBy while customer_id is still present
    strag_salted.filter(col("customer_id").isin(stragglers)) \
        .groupBy("salted_customer_id") \
        .count() \
        .orderBy(col("salted_customer_id").desc()) \
        .show()


    # Stage 1: partial aggregates per salt bucket — spreads the VIP customers'
    # hot rows across 10 synthetic keys instead of 1 real key
    partial_agg = strag_salted.groupBy("salted_customer_id").agg(
                  first("customer_id").alias("customer_id"),
                  count_distinct("order_id").alias("partial_order_count"),
                  sum("quantity").alias("partial_total_quantity"),
                  sum(col("quantity") * col("unit_price")).alias("partial_total_value")
              )
    # Count occurrences of each (customer_id, region) pair
    customer_region_counts = salt_df.groupBy("customer_id", "region").count()


    # Rank regions per customer by frequency descending
    window_spec = Window.partitionBy("customer_id").orderBy(col("count").desc())

    customer_mode_region = (
    customer_region_counts
    .withColumn("rank", row_number().over(window_spec))
    .filter(col("rank") == 1)
    .select("customer_id", col("region").alias("primary_region"))
    )

    # Stage 2: combine the 10 partials per real customer_id back into one row
    final_agg = partial_agg \
                 .groupBy("customer_id").agg(
                 sum("partial_order_count").alias("order_count"),
                 sum("partial_total_quantity").alias("total_quantity"),
                 round(sum("partial_total_value"), 2).alias("total_order_value")
                 ).join(customer_mode_region, on="customer_id", how="left") \
                  .withColumn("avg_order_value", round((col("total_order_value") / col("order_count")), 2)) \

    # Explain all (salted) agg in gold layer
    explain_and_log(partial_agg, "Gold – salted aggregation")

    final_agg.write.format("delta") \
              .mode("overwrite").option("overwriteSchema", "true") \
              .partitionBy("primary_region").save(agg_path)

    # Z-Orderinng for gold writen to delta
    DeltaTable.forPath(spark, agg_path).optimize().executeZOrderBy("avg_order_value")

gold_layer(silver_df, gold_path)
t7 = time.time()
print(f"Gold-salted-agg & Skew-Audit Run_time: {t7 - t6:.2f}s")

print("Gold salted history:")
DeltaTable.forPath(spark, gold_path).history().show(truncate=False)

silver_df.unpersist()

gold_df = spark.read.format("delta").load(gold_path)
gold_df.show(10, truncate=False)

size11 = dataframe_size(silver_df)
print(f"gold_df size: {size11 / (1024):.2f} KiB")

# Mlops
# import Mlops components & utility modules
from pyspark.ml import Pipeline
from pyspark.ml.feature import (StringIndexer,
                                OneHotEncoder, VectorAssembler)
from pyspark.ml.regression import (LinearRegression,
                                   RandomForestRegressor)
from pyspark.ml.evaluation import RegressionEvaluator
import mlflow
import mlflow.spark
from pyspark.ml import PipelineModel

def ml_layer(gold_df, ml_df, model_path, delta_ml):
  """
      * gold_df = gold_df,
        ml_df = good_ml,
        model_path = model_path
        delta_ml = delta_ml*
  """

  # Required data for training
  ml_data_1 = gold_df.select("customer_id",
                             "primary_region",
                             "order_count",
                             "total_quantity",
                             "total_order_value",
                             "avg_order_value")

  # Extract VIPs(stragglers) before Train-Test Split to avoid inbalance
  vip_ids = ["9001", "9002", "9003", "9004", "9005"]
  vip_df = ml_data_1.filter(col("customer_id").isin(vip_ids))
  normal_df = ml_data_1.filter(~col("customer_id").isin(vip_ids))

  # Train-Test Split
  normal_train, normal_test = normal_df.randomSplit([0.8, 0.2], seed=42)
  train_df = normal_train.union(vip_df)   # VIPs always in training
  test_df = normal_test                    # test set stays "normal-customer" only

  # MLOps Experiment Tracking & Pipeline Definition
  mlflow.set_experiment("Delta-PySpark_Regression")

  with mlflow.start_run():
    # Index the text column
    region_indexer = StringIndexer(inputCol="primary_region",
                                   outputCol="region_indexed")

    # One-hot encode the indexed region
    region_encoder = OneHotEncoder(inputCol="region_indexed",
                                   outputCol="region_encoded")

    # Assemble into a vector
    assembler = VectorAssembler(
    inputCols=["region_encoded",
                 "order_count",
                 "total_quantity",
                 "avg_order_value"
                 ],
              outputCol="features",
              handleInvalid="skip")

    train_df = train_df.withColumn("log_total_order_value", log1p(col("total_order_value")))
    test_df = test_df.withColumn("log_total_order_value", log1p(col("total_order_value")))

    # Random Forest Regressor Parameters
    rf_regressor = RandomForestRegressor(
    featuresCol="features",
    labelCol="log_total_order_value",
    predictionCol="log_prediction",
    numTrees=100,              # Number of trees in the forest
    maxDepth=10,               # Max depth per tree (regularization)
    minInstancesPerNode=5,     # Min data points required to split a node (regularization)
    featureSubsetStrategy="auto", # Random features chosen per split
    seed=42
    )

    # Construct Spark Pipeline
    pipeline = Pipeline(stages=[region_indexer,
                                region_encoder,
                                assembler,
                                rf_regressor])

    # Fit Pipeline
    model = pipeline.fit(train_df)

    # Make Predictions on Test Set
    predictions = model.transform(test_df)

    model.write().overwrite().save(model_path)

    predictions = predictions.withColumn("prediction",
                                         expm1(col("log_prediction")))  # back to real scale

    print("\n--- Model Predictions Sample ---")
    predictions.select("customer_id",
                       "primary_region",
                       "total_order_value",  # Actual target label
                       "prediction"         # Predicted model output
                       ).show(10, truncate=False)

    # Evaluate Model
    evaluator_rmse = RegressionEvaluator(labelCol="total_order_value",
                                         predictionCol="prediction", metricName="rmse")
    evaluator_r2 = RegressionEvaluator(labelCol="total_order_value",
                                       predictionCol="prediction", metricName="r2")

    rmse = evaluator_rmse.evaluate(predictions)
    r2 = evaluator_r2.evaluate(predictions)

    # Log Metrics to MLOps Platform
    mlflow.log_metric("rmse", rmse)
    mlflow.log_metric("r2", r2)

    # Log Pipeline Artifact (Model Registry Ready)
    mlflow.spark.log_model(model, artifact_path="spark-regression-model")

    print(f"Model Trained Successfully. Test RMSE: {rmse:.2f}, R2: {r2:.2f}")

    normal_test_preds = predictions.filter(~col("customer_id").isin(vip_ids))
    print("Normal-customer RMSE/R²:", evaluator_rmse.evaluate(normal_test_preds),
          evaluator_r2.evaluate(normal_test_preds))

    # Count occurrences of each (customer_id, region) pair
    customer_region_counts = ml_df.groupBy("customer_id", "region").count()

    # Rank regions per customer by frequency descending
    window_spec = Window.partitionBy("customer_id").orderBy(col("count").desc())

    customer_mode_region = (
    customer_region_counts
    .withColumn("rank", row_number().over(window_spec))
    .filter(col("rank") == 1)
    .select("customer_id", col("region").alias("primary_region"))
    )

    ml_agg = ml_df.groupBy("customer_id").agg(
                  count_distinct("order_id").alias("order_count"),
                  sum("quantity").alias("total_quantity"),
                  sum(col("quantity") * col("unit_price")).alias("total_order_value")
              ).join(customer_mode_region, on="customer_id", how="left") \
                  .withColumn("avg_order_value", round((col("total_order_value") / col("order_count")), 2))

    # Explain ml_agg in ml_layer
    explain_and_log(ml_agg, "ml_agg - aggregation")

    # Score the new batch **no retraining**
    score_model = model.transform(ml_agg)
    score_model = score_model.withColumn("prediction", expm1(col("log_prediction")))

    # 2. Display sample predictions
    print("\n--- Scored Batch Data Sample ---")
    score_model.select("customer_id", "primary_region", "prediction").show(5, truncate=False)

    score_model.write.format("delta") \
              .mode("overwrite") \
              .option("overwriteSchema", "true") \
              .partitionBy("primary_region").save(delta_ml)

    # Z-Orderinng for ml writen to delta
    DeltaTable.forPath(spark, delta_ml).optimize().executeZOrderBy("avg_order_value")

gold_df.cache()
good_ml.cache()

t8 = time.time()
ml_layer(gold_df=gold_df,
         ml_df=good_ml,
         model_path=model_path,
         delta_ml=delta_ml)

print("ML score_model history:")
DeltaTable.forPath(spark, delta_ml).history().show(truncate=False)

gold_df.unpersist()
good_ml.unpersist()

t9 = time.time()
print(f"ML Run_time: {t9 - t8:.2f}s")

ml_load_df = spark.read.format("delta").load(delta_ml)

size11 = dataframe_size(ml_load_df)
print(f"ml_load_df size: {size11 / (1024):.2f} KiB")

print("\nPipeline finished. All caches released.")
print("Open Spark UI → http://localhost:4050 to review jobs, stages, SQL plans and storage.")

# File counts and size after Z-ORDER
print("=== File counts after Z-ORDER ===")
for path, name in [
    (bronze_path1, "Bronze_df1"),
    (bronze_path2, "Bronze_df2"),
    (ml_path, "ml_df"),
    (silver_path, "Silver_df"),
    (gold_path, "Gold_df"),
    (delta_ml, "ML-score_model")
]:
    detail = DeltaTable.forPath(spark, path).detail().select("numFiles", "sizeInBytes").collect()[0]
    print(f"{name}: {detail['numFiles']} files, {detail['sizeInBytes']/1024:.1f} KiB")
```
